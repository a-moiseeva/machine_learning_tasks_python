```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

from time import time
```

```python
db_train = pd.read_csv("holdout_data/train.csv")
db_test = pd.read_csv("holdout_data/holdout.csv")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.model_selection import cross_val_score
from sklearn.metrics import roc_auc_score

from sklearn.ensemble import RandomForestRegressor

from sklearn.model_selection import GridSearchCV
```

```python
target_col = "raintomorrow"
id_cols = ["row_id"]
```

```python
db_test.drop(columns=id_cols, inplace = True)
db_train.drop(columns=id_cols, inplace = True)
```

```python
numeric_cols = [x for x in db_train.select_dtypes(include=np.number).columns.tolist() if x not in id_cols]
cat_cols = [x for x in db_train.columns if x not in numeric_cols and x not in id_cols]
```

```python
db_train_with_dummies = pd.get_dummies(db_train, columns=cat_cols)
db_test_with_dummies = pd.get_dummies(db_test, columns=cat_cols)
```

#### Предобработаем данные - заполним отстуствующие значения.

Сделаем это с помощью `RandomForestRegressor`. Для каждого столбца, в котором есть пропущенные значения будем делить наш изначальный тренировочный сет на два новых временных: тренировочный (без отсутсвующих значений в данном столбце) и на тестовый - с пропущенными: будем обучать модель, строить предсказания и заполнять ими отстствующие значения.

Интересное наблюдение: когда делаем регрессию для заполнения нанов в конкретном столбце, у нас в начале все еще есть незаполненные наны в других столбцах. Их можно запонять средним/медианой или банально -999. Оказывается, что второй вариант намного лучше.

```python
for target in db_train_with_dummies.columns:
    if (db_train_with_dummies[target].isna().sum()>0):
        df_train = db_train_with_dummies[db_train_with_dummies[target].isna() == 0]
        df_test = db_train_with_dummies[db_train_with_dummies[target].isna() == 1]

        df_test_2 = db_test_with_dummies[db_test_with_dummies[target].isna() == 1]


        y_train = df_train[target].values
        df_train.drop(columns=[target], inplace=True)
        df_test.drop(columns=[target], inplace=True)
        df_test_2.drop(columns=[target], inplace=True)

        x_train = df_train.fillna(-9999).values
        x_test = df_test.fillna(-9999).values
        x_test_2 = df_test_2.fillna(-9999).values

        regr = RandomForestRegressor(n_estimators = 1500, n_jobs = -1, random_state = 36)
        regr.fit(x_train, y_train)

        db_train_with_dummies.loc[db_train_with_dummies[target].isna(), target] = regr.predict(x_test)
        db_test_with_dummies.loc[db_test_with_dummies[target].isna(), target] = regr.predict(x_test_2)
```

```python
Y_train = db_train_with_dummies[target_col].values
db_train_with_dummies.drop(columns=[target_col], inplace=True)

Y_test = db_test_with_dummies[target_col].values
db_test_with_dummies.drop(columns=[target_col], inplace=True)
```

Посмотрим результаты

```python
X_train = db_train_with_dummies.values
X_test = db_test_with_dummies.values
```

```python
clf = RandomForestClassifier(n_estimators = 1500, n_jobs = -1, random_state = 42)

clf.fit(X_train, Y_train)
preds = clf.predict_proba(X_test)[:, 1]
roc_auc_score(Y_test, preds)
```
